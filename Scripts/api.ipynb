```python
import archiver_api as arc
import pandas as pd
import os
import datetime
```

```python
olog_data = pd.read_csv("../Data/olog_data/olog_data.csv")
olog_data
```

```python
for i, log in olog_data.iterrows():
    filename = f"../Data/sismic_data/{log['id']}.xlsx"
    
    print(f"LOG {i:2} - {log['id']}: Procurando dados...")

    try:
        with pd.ExcelWriter(filename, engine="openpyxl") as writer:
            
            for axis in ["Z", "X", "Y"]:
                
                from_time = arc.sub_hours(log["from"], 1)
                to_time = arc.add_hours(log["to"], 0.5)

                delta = int(abs(arc.seconds_between(from_time, to_time)))
                
                df = arc.get_archiver_data(
                    pv=f"RR-09S:SS-Seism-Ax13:S{axis}-Mon",
                    from_time=from_time,
                    to_time=to_time,
                    optimize=delta
                )
                
                df.to_excel(
                    writer,
                    sheet_name=f"S{axis}",
                    index=False
                )
            
            for axis in ["X", "Y"]:
                        
                        from_time = arc.sub_hours(log["from"], 2)
                        to_time = arc.add_hours(log["to"], 2)
                        
                        df = arc.get_archiver_data(
                            pv=f"SI-01C1:DI-BPM-1:Pos{axis}-Mon",
                            from_time=from_time,
                            to_time=to_time,
                            optimize=delta
                        )
                        
                        df.to_excel(
                            writer,
                            sheet_name=f"BPM{axis}",
                            index=False
                        )
                        
                
        print(f"LOG {i:2} - {log["id"]}: Processamento Concluído!")
    except:
        os.remove(filename)
        print(f"LOG {i:2} - {log["id"]}: Erro no processamento")
```
